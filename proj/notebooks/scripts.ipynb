## Compose the TSV file from ASV data file

```python
import pandas as pd

ASV_PROTOCOL_FILE = '../data/asv2019_la/ASVspoof2019_LA_cm_protocols/ASVspoof2019.LA.cm.eval.trl.txt'
PROTOCOL_COLS = ['speaker_id', 'audio_file_name', 'system_id', 'xxx', 'key']
```

```python
asv_protocol = pd.read_csv(ASV_PROTOCOL_FILE, sep=' ', names=PROTOCOL_COLS)
print(asv_protocol.columns)
print(len(asv_protocol))
print(asv_protocol.iloc[42])
```

```python
asv_protocol['audio_file_name'] = asv_protocol['audio_file_name'] + '.flac'
print(asv_protocol.iloc[42])
```

```python
asv_protocol.to_csv('asv_2019.eval.tsv', sep='\t', index=False)
```
